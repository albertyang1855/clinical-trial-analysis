```python
import pandas as pd
print(f"pandas: {pd.__version__}")


```

```python
# a DataFrame is a table
pd.DataFrame({'column1':[1,0,0],
              'column2':[2,0,0]},
            index=['Product A', 'Product B','Product C'])
```

```python
# a table with one column. a default column name '0' is assigned
pd.DataFrame({2,4,6},
             index=['Product A', 'Product B','Product C'])
```

```python
# a Series is a list
pd.Series([1,3,5,7,9])
```

```python
# a Series is a list can also use index. Series do not use column name. 
pd.Series([1,3,5], 
          index=['Product A', 'Product B','Product C'],
         name='Product')
```
